```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt


rho = 2710.0        # Density (kg/m^3)
r = 0.1             # Disk radius (m)
h = 0.01            # Disk thickness (m)
d = 0.05            # Slot offset (m)
b = 0.01            # Half-width of slot (m)
l = 0.08            # Half-length of slot (m)
g = 10.0            # Gravity (m/s^2)
m = 0.05            # Bead mass (kg) - Assumed
k = 10.0            # Spring constant (N/m) - Assumed


M_full = rho * np.pi * r**2 * h
M_slot = rho * (4 * l * b * h)
M_disk = M_full - M_slot

I11_full = M_full * (3*r**2 + h**2) / 12
I33_full = 0.5 * M_full * r**2

I11_cut = M_slot * ((2*b)**2 + h**2) / 12
I22_cut = M_slot * ((2*l)**2 + h**2) / 12
I33_cut = M_slot * ((2*l)**2 + (2*b)**2) / 12


I1 = I11_full - (I11_cut + M_slot * d**2)
I2 = I11_full - I22_cut
I3 = I33_full - (I33_cut + M_slot * d**2)


def eom(t, state):
    y, theta, phi, y_dot, theta_dot, phi_dot = state


    M = np.zeros((3, 3))
    M[0, 0] = m
    M[0, 2] = -m * d

    I_eff = I1 * np.cos(phi)**2 + I2 * np.sin(phi)**2 + m * (d * np.cos(phi) + y * np.sin(phi))**2
    M[1, 1] = I_eff

    M[2, 0] = -m * d
    M[2, 2] = I3 + m * d**2 + m * y**2


    F = np.zeros(3)

    # y-force
    F[0] = m * y * phi_dot**2 + m * theta_dot**2 * np.sin(phi) * (d * np.cos(phi) + y * np.sin(phi)) - k * y - m * g * np.sin(theta) * np.sin(phi)

    # theta-force
    I_eff_dot = 2 * (I2 - I1) * np.sin(phi) * np.cos(phi) * phi_dot + \
                2 * m * (d * np.cos(phi) + y * np.sin(phi)) * (y_dot * np.sin(phi) + (y * np.cos(phi) - d * np.sin(phi)) * phi_dot)
    F[1] = -I_eff_dot * theta_dot - g * np.cos(theta) * (m * y * np.sin(phi) + (m - M_slot) * d * np.cos(phi))

    # phi-force
    dL_dphi = (I2 - I1) * theta_dot**2 * np.sin(phi) * np.cos(phi) + \
              m * theta_dot**2 * (d * np.cos(phi) + y * np.sin(phi)) * (-d * np.sin(phi) + y * np.cos(phi)) - \
              g * np.sin(theta) * (m * y * np.cos(phi) - (m - M_slot) * d * np.sin(phi))
    F[2] = -2 * m * y * y_dot * phi_dot + dL_dphi

    # Solve M * q_ddot = F
    accel = np.linalg.solve(M, F)

    return [y_dot, theta_dot, phi_dot, accel[0], accel[1], accel[2]]

state0 = [0.0, np.pi/4, 0.0, 0.0, 0.0, 15.0]
t_span = (0, 5)
t_eval = np.linspace(t_span[0], t_span[1], 2000)

sol = solve_ivp(eom, t_span, state0, t_eval=t_eval, method='RK45', rtol=1e-8, atol=1e-8)

plt.figure(figsize=(10, 8))

plt.subplot(3, 1, 1)
plt.plot(sol.t, sol.y[0, :], 'b', label='$y(t)$')
plt.ylabel('Bead Position (m)')
plt.legend(loc='upper right')
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(sol.t, np.degrees(sol.y[1, :]), 'r', label='$\\theta(t)$')
plt.ylabel('Pitch Angle (deg)')
plt.legend(loc='upper right')
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(sol.t, np.degrees(sol.y[2, :]), 'g', label='$\\phi(t)$')
plt.ylabel('Spin Angle (deg)')
plt.xlabel('Time (s)')
plt.legend(loc='upper right')
plt.grid(True)

plt.tight_layout()
plt.show()
```
